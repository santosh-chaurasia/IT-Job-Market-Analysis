**Description:**
This notebook generates a simulated dataset of 26,500 IT job postings and prepares it for analysis. The data is not real hiring data and was created for practice.

# ===================================================
# STEP 1: DATA GENERATION
# Purpose: Creating a large dataset of 26,500+ rows to analyze IT job trends


```python
import random
import pandas as pd

# 1. Define lists representing the Indian IT industry
companies = [
    "TCS",
    "Infosys",
    "Wipro",
    "Cognizant",
    "Accenture",
    "Tech Mahindra",
    "Capgemini",
    "JPMorgan",
    "Amazon India",
    "Allied Digital",
    "eClerx",
    "Parascadd",
]

cities = [
    "Navi Mumbai",
    "Mumbai",
    "Bengaluru",
    "Pune",
    "Hyderabad",
    "Noida",
    "Chennai",
]

experience_levels = ["Entry-Level", "Mid-Level", "Senior"]
years = [2024, 2025, 2026]

# 2. Separate job roles into AI/Data categories and Traditional categories
ai_data_roles = ["Data Analyst", "AI / ML Engineer", "Data Engineer"]
traditional_roles = [
    "Java Developer",
    "Manual Tester",
    "System Administrator",
    "UI Developer",
    "Python Developer",
]

# Create an empty list to store all generated job rows
all_jobs_data = []
total_rows = 26500

# 3. Start a loop to generate data row by row
for i in range(total_rows):
    # Generate a unique Job ID
    job_id = "JOB_" + str(100000 + i)

    # Pick random values from our lists
    company = random.choice(companies)
    location = random.choice(cities)
    exp_level = random.choice(experience_levels)
    year = random.choice(years)

    # Market Shift Logic: Traditional jobs drop in 2026, while AI/Data jobs increase
    if year == 2024:
        role_type = random.choices(["trad", "ai"], weights=[70, 30])[0]
    elif year == 2025:
        role_type = random.choices(["trad", "ai"], weights=[50, 50])[0]
    else:  # Year 2026
        role_type = random.choices(["trad", "ai"], weights=[30, 70])[0]

    # Assign Job Titles and Skills based on the selected role type
    if role_type == "ai":
        title = random.choice(ai_data_roles)
        if title == "Data Analyst":
            skills = "SQL, Python, Power BI, Advanced Excel, Data Analysis"
        elif title == "AI / ML Engineer":
            skills = "Python, Machine Learning, LLM, Prompt Engineering, NLP"
        else:
            skills = "AWS, Spark, SQL, Python, ETL Pipelines"
    else:
        title = random.choice(traditional_roles)
        if title == "Java Developer":
            skills = "Java, Spring Boot, Hibernate, SQL, Git"
        elif title == "Manual Tester":
            skills = "Manual Testing, Test Cases, Jira, Excel"
        elif title == "UI Developer":
            skills = "HTML, CSS, JavaScript, React JS, Tailwind CSS"
        elif title == "Python Developer":
            skills = "Python, Django, Flask, SQL, REST APIs"
        else:
            skills = "Linux, Windows Server, Networking, IT Support"

    # Salary Logic (INR) based on Experience Level and Role Type
    if exp_level == "Entry-Level":
        if role_type == "ai":
            salary = random.randint(350000, 600000)
        else:
            salary = random.randint(240000, 400000)

    elif exp_level == "Mid-Level":
        if role_type == "ai":
            salary = random.randint(700000, 1200000)
        else:
            salary = random.randint(500000, 850000)

    else:  # Senior level
        if role_type == "ai":
            salary = random.randint(1400000, 2200000)
        else:
            salary = random.randint(950000, 1500000)

    # Give a 15% salary hike to AI roles specifically in 2026 due to high demand
    if year == 2026 and role_type == "ai":
        salary = int(salary * 1.15)

    # Collect all features into a single row list
    single_row = [
        job_id,
        company,
        title,
        skills,
        exp_level,
        salary,
        location,
        year,
    ]

    # Append the row to our main list
    all_jobs_data.append(single_row)

# 4. Convert the full list into a Pandas DataFrame
columns_list = [
    "Job_ID",
    "Company_Name",
    "Job_Title",
    "Skills_Required",
    "Experience_Level",
    "Salary_INR",
    "Location",
    "Posted_Year",
]
df = pd.DataFrame(all_jobs_data, columns=columns_list)

# 5. Save the generated data 
df.to_csv("simulated_it_jobs.csv", index=False)

# 6. Confirm the data generation is complete
print("Dataset generated successfully.")
print("Total rows generated:", len(df))
```

# ============================================================
# STEP 3: DATA TRANSFORMATION FOR SQL NORMALIZATION
# Purpose: Splitting the comma-separated skills so each skill gets its own row



```python

# 1. Create an empty list to collect individual skill rows
skills_data_list = []

# 2. Loop through our main dataframe to split the skills string
for index, row in df.iterrows():
    current_job_id = row["Job_ID"]
    # Split the skills string by comma
    skills_list = row["Skills_Required"].split(",")

    # Clean up any extra spaces around each skill name and save it
    for skill in skills_list:
        clean_skill = skill.strip()
        skills_data_list.append([current_job_id, clean_skill])

# 3. Convert the list into a specialized normalized DataFrame
df_skills_normalized = pd.DataFrame(
    skills_data_list, columns=["Job_ID", "Skill"]
)

# 4. Save this new normalized table to a separate CSV file for SQL import
df_skills_normalized.to_csv("Normalized_Job_Skills.csv", index=False)

# 5. Print confirmation and see how it transformed
print("Transformation Complete!")
print(f"Total Normalized Skill Rows: {len(df_skills_normalized)}")
print("\nFirst 10 rows of your new SQL-ready Skills Table:")
print(df_skills_normalized.head(10))
```
