```python
import pandas as pd
import plotly.express as px

# Step 1: Load the cleaned CSV file directly
df = pd.read_csv('IT_Job_Market_Clean.csv')

# Step 2: Calculate the average salary for each job title
job_data = df.groupby('Job_Title')['Salary_INR'].mean().reset_index()

# Step 3: Sort and get the top 10 highest-paying jobs
top_jobs = job_data.sort_values(by='Salary_INR', ascending=False).head(10)

# Step 4: Create a simple interactive horizontal bar chart
fig = px.bar(
    top_jobs,
    x='Salary_INR',
    y='Job_Title',
    orientation='h',
    title='Top 10 Highest Paying Job Titles',
    labels={'Salary_INR': 'Average Salary (INR)', 'Job_Title': 'Job Title'},
    color='Salary_INR'
)

# Step 5: Display the chart
fig.show()
```

```python
# Step 2: Count the total number of jobs for each location
location_counts = df['Location'].value_counts().reset_index()
location_counts.columns = ['Location', 'Job_Count']

# Step 3: Create a clean vertical bar chart for job counts
fig2 = px.bar(
    location_counts,
    x='Location',
    y='Job_Count',
    title='Total Job Openings by Location',
    labels={'Location': 'City', 'Job_Count': 'Number of Openings'},
    color='Job_Count'
)

# Step 4: Display the chart
fig2.show()
```
